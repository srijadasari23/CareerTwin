## 5. Launch Full-Stack Server

Starts the Express + React full-stack application on port 3000.

```python
import subprocess, time, urllib.request

server_proc = subprocess.Popen(
    ["npm", "run", "dev"],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)

print("Waiting for CareerTwin AI server on port 3000...", end="", flush=True)
is_up = False
for _ in range(30):
    try:
        res = urllib.request.urlopen("http://localhost:3000/api/health", timeout=2)
        if res.status == 200:
            is_up = True
            break
    except Exception:
        print(".", end="", flush=True)
        time.sleep(1)

if is_up:
    print("\n✅ CareerTwin AI backend & frontend are active on http://localhost:3000")
else:
    print("\n⚠️ Server starting in background. Check terminal logs.")
```
